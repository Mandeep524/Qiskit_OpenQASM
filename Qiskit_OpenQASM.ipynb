# Modified 3-Qubit Circuit with QASM Export & Simulation

```

from qiskit import QuantumCircuit, Aer, execute
from qiskit.visualization import plot_histogram

# Create a 3-qubit circuit
qc = QuantumCircuit(3, 3)

# Apply custom gate sequence
qc.h(0)            # Hadamard on qubit 0
qc.cx(0, 1)        # CNOT from qubit 0 to 1
qc.rx(1.2, 2)      # Rotation gate on qubit 2
qc.barrier()       # Add barrier to separate stages
qc.cx(1, 2)        # CNOT from qubit 1 to 2
qc.h(2)            # Hadamard on qubit 2

# Measure all qubits
qc.measure([0,1,2], [0,1,2])

qc.draw('mpl')
```

## Generate and Display QASM Code

```

qasm_code = qc.qasm()
print(qasm_code)
```


### **How Gates Appear in QASM**
| Operation | QASM Syntax |
|----------|-------------|
| Hadamard (H) | `h q[x];` |
| CNOT (CX) | `cx q[a], q[b];` |
| Barrier | `barrier q[x], q[y], ...;` |


## Run the Circuit Simulation

```

backend = Aer.get_backend('qasm_simulator')
result = execute(qc, backend, shots=1024).result()
counts = result.get_counts()

plot_histogram(counts)
```
